```python
import shutil
from pathlib import Path

cands = list(Path('/kaggle/input').glob('*/processed/train.jsonl'))
assert cands, 'Liga o dataset football-laya-curve em Add data!'
SRC = cands[0].parents[1]
DST = Path('/kaggle/working/fb')
for name, dest in (('processed', DST / 'data' / 'processed'),
                   ('models', DST / 'data' / 'models'),
                   ('scripts', DST / 'scripts')):
    if dest.exists():
        shutil.rmtree(dest)
    shutil.copytree(SRC / name, dest)
print('tree pronta:', sorted(str(p.relative_to(DST)) for p in DST.rglob('*') if p.is_file())[:8], '...')
```
